# Datasets for Helsinki Metropolitan Area


## Helsinki administrative borders

```python
import geopandas as gpd

borders = gpd.read_file("Helsinki/Helsinki_borders.shp")
borders.head()
```

```python
borders.explore()
```

## Travel times to central railway station, Helsinki

```python
import geopandas as gpd

ttimes = gpd.read_file("Helsinki/TravelTimes_to_5975375_RailwayStation.shp")
ttimes.head()
```

```python
ttimes.explore("pt_r_tt", cmap="RdYlBu", scheme="naturalbreaks", k=6)
```

## Metro line

## Train and metro stops

```python
import geopandas as gpd

stops = gpd.read_file("Helsinki/addresses.shp")
stops.head()
```

```python
stops.explore(color="red", marker_kwds=dict(radius=4))
```
