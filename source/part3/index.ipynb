# Overview and learning goals

This part of the book provides a few examples of what can be done building on the skills you have learned in the first two parts of the book. These "case studies" are a bit more advanced than earlier examples in the book, but are also closer to the types of Python-based work that would be used in research or professional data analysis. This part also provides a summary of what you have learned and some ideas to consider in your future uses of Python for geographic data analysis.

```{admonition} Learning goals
:class: tip

At the end of this part, you should be able to:

- Generate travel time matrices by different travel modes (public transport, cycling, car)
- Understand the basic concepts related to watershed analysis
- Analyze various topographic factors influencing watersheds
- See how Python fits within the realm of other programming languages for geographic data analysis
- Understand the challenges and benefits of the rapidly evolving Python ecosystem
```
